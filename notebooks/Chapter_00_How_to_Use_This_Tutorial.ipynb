## 0.1 What is a Jupyter Notebook?

A Jupyter Notebook is a document that mixes two things:
- **Text** — explanations, instructions, examples (like what you are reading now)
- **Code** — Python programs you can run directly in the browser

Each block is called a **cell**. There are two types:

| Cell Type | What it contains | Background colour |
|---|---|---|
| Markdown cell | Text, headings, tables | White |
| Code cell | Python code | Light grey |

The cell below is a **code cell**. You can see it has a grey background and a `[ ]` on the left side.

````python
print("This is a code cell")
````

## 0.5 Run Cells in Order

**Important:** Always run cells from top to bottom, in order.

Each cell can use variables and functions defined in cells above it. If you skip a cell or run them out of order, you will get errors.

Example — run these two cells in order:

````python
# Cell 1 — defines a variable
temperature = 32
````

````python
# Cell 2 — uses the variable from Cell 1
print(f"Temperature is {temperature}°C")   # works only if Cell 1 ran first
````

If you run Cell 2 without running Cell 1 first, you will see:  
`NameError: name 'temperature' is not defined`

The fix: run Cell 1 first, then Cell 2.

## 0.6 What to Do When Something Goes Wrong

### The cell shows an error
Read the last line of the error message — it tells you what went wrong. Fix the code and run the cell again.

### The cell is stuck (the `[*]` never changes to a number)
This means the code is still running or is stuck in a loop. To stop it:  
**Kernel → Interrupt** (or press the ■ stop button in the toolbar)

### Everything seems broken / variables have wrong values
Restart the kernel and run all cells from the top:  
**Kernel → Restart & Run All**

This resets everything to a clean state.

````python
# This cell has an error on purpose — read the error message
print("This line works")
print(undefined_variable)   # NameError — this variable was never defined
````

## 0.7 The Toolbar

The toolbar at the top of the notebook has the most common actions:

| Button | What it does |
|---|---|
| 💾 (floppy disk) | Save the notebook |
| ＋ | Add a new cell below |
| ✂️ | Cut the selected cell |
| ▶ Run | Run the selected cell |
| ■ Stop | Stop a running cell |
| ⟳ Restart | Restart the kernel (clears all variables) |
| ⏩ Run All | Restart and run every cell from top to bottom |

You will use **▶ Run**, **■ Stop**, and **⟳ Restart** most often.

## 0.8 Saving Your Work

This is the most important section if you are using **Binder** or **Colab**.

---

### If you are using Binder

```{warning}
Binder does NOT save your work automatically. If you close the tab or the session times out (after ~10 minutes of inactivity), everything you typed is lost.
```

**To save your work before closing:**

1. From the menu: **File → Download as → Notebook (.ipynb)**
2. Save the downloaded file to your computer
3. Next time you open Binder, upload the file: **File → Open** → drag and drop your saved file

**Do this at the end of every study session.**

---

### If you are using Google Colab

```{warning}
Colab opens a temporary copy of the notebook. Changes are NOT saved back to GitHub.
```

**To save your own copy before you start:**

1. From the menu: **File → Save a copy in Drive**
2. A copy opens in your Google Drive — work from that copy
3. Colab auto-saves to your Drive from that point on

**Do this as soon as you open a chapter in Colab.**
